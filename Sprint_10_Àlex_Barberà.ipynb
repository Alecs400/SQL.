Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:

- Llista de publicacions (/posts)
- Llista d’usuaris (/users)
- Llista de tasques (/todos)

```python
import requests

# Definimos las URLs de la API de JSONPlaceholder
endpoints = {
    "Publicaciones": "https://jsonplaceholder.typicode.com/posts",
    "Usuarios": "https://jsonplaceholder.typicode.com/users",
    "Tareas": "https://jsonplaceholder.typicode.com/todos"
}
```

2. Mostra per pantalla:

- La quantitat total de cada recurs.
- El codi d'estat de cada petició.

```python
# Realizamos las peticiones y mostramos los resultados
for nombre, url in endpoints.items():
    response = requests.get(url)

    print(f"Recurso: {nombre}")
    print(f" - Código de estado: {response.status_code}")

    if response.status_code == 200:
        datos = response.json()
        print(f" - Cantidad total: {len(datos)}")
    else:
        print(" - Error al obtener los datos.")
    print("-" * 30)
```

4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
import requests

# URL del endpoint para crear publicaciones
url = "https://jsonplaceholder.typicode.com/posts"

# Datos que queremos enviar (título, cuerpo y userId)
nuevo_post = {
    "title": "Mi nueva publicación de prueba",
    "body": "Este es el contenido del mensaje para la práctica de requests.",
    "userId": 1
}

# Realizamos la petición POST enviando los datos con el parámetro json
response = requests.post(url, json=nuevo_post)

# Mostramos el código de estado
print(f"Código de estado: {response.status_code}")

# Mostramos la respuesta JSON recibida del servidor
print("Respuesta JSON:")
print(response.json())
```

5. Fes una petició PATCH per modificar parcialment una publicació existent.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
import requests

# URL del endpoint para modificar una publicación existente (por ejemplo, el post con id 1)
url = "https://jsonplaceholder.typicode.com/posts/1"

# Datos parciales que queremos actualizar
datos_actualizados = {
    "title": "Nuevo título actualizado parcialmente con PATCH"
}

# Realizamos la petición PATCH enviando los datos con el parámetro json
response = requests.patch(url, json=datos_actualizados)

# Mostramos el código de estado
print(f"Código de estado: {response.status_code}")

# Mostramos la respuesta JSON recibida del servidor
print("Respuesta JSON:")
print(response.json())
```

6. Fes una petició DELETE sobre una publicació.

- Mostra la resposta JSON.
- Mostra el codi d'estat.


```python
import requests

# URL del endpoint para eliminar una publicación existente (por ejemplo, el post con id 1)
url = "https://jsonplaceholder.typicode.com/posts/1"

# Realizamos la petición DELETE
response = requests.delete(url)

# Mostramos el código de estado (acostumbra a ser 200 OK)
print(f"Código de estado: {response.status_code}")

# Mostramos la respuesta JSON recibida del servidor (en JSONPlaceholder suele ser un diccionario vacío {})
print("Respuesta JSON:")
print(response.json())
```

3. Fes una petició GET senzilla:

- Mostra el codi d'estat de la resposta.
- Imprimeix de forma clara alguns dels camps de la resposta JSON.


```python
import requests

# URL del endpoint para obtener un dato curioso aleatorio
url = "https://catfact.ninja/fact"

# Realizamos la petición GET a la API
response = requests.get(url)

# Mostramos el código de estado de la respuesta
print(f"Código de estado: {response.status_code}")

# Comprobamos si la petición ha sido exitosa (200 OK)
if response.status_code == 200:
    # Convertimos la respuesta a un diccionario JSON
    datos = response.json()

    # Imprimimos de forma clara los campos obtenidos
    print("\nDatos de la respuesta JSON:")
    print(f"• Dato curioso (fact): {datos.get('fact')}")
    print(f"• Longitud (length): {datos.get('length')}")
else:
    print("Error al realizar la petición.")
```

4. Converteix les dades a un DataFrame de pandas:

- Mostra les primeres files del DataFrame.

```python
import pandas as pd
import requests

# URL del endpoint para obtener una lista de datos curiosos (para tener múltiples filas)
url = "https://catfact.ninja/facts"

# Realizamos la petición GET
response = requests.get(url)

# Comprobamos que la petición ha sido correcta
if response.status_code == 200:
    # Obtenemos la respuesta en formato JSON
    data = response.json()

    # La lista de hechos se encuentra dentro de la clave 'data'
    llista_fets = data["data"]

    # Convertimos la lista de diccionarios a un DataFrame de pandas
    df = pd.DataFrame(llista_fets)

    # Mostramos las primeras filas del DataFrame
    print("Primeras filas del DataFrame:")
    print(df.head())
else:
    print(f"Error al realizar la petición: {response.status_code}")
```
