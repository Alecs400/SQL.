#Funciones

Paso 1: Entrada y extracción: obtenir_dades (prepara el terreno).

Paso 2: Control de calidad: validar_dades (filtro de seguridad).

Paso 3: Lógica de negocio: calcular_puntuacio (aplica las reglas de la empresa).

Paso 4: Toma de decisiones: classificar_candidat (traduce resultados a conclusiones).


```python
def obtenir_dades():
    print("\n--- Formulario de inscripción de candidatos ---")

    # 1. Bucle para validar obligatoriamente que el nombre sea correcto (solo letras y espacios)
    while True:
        nom = input("Introduce el nombre del candidato: ").strip()

        if nom.lower() == 'salir':
            return ['salir', '', '', '', '', '']

        if not nom:
            print("Error: El nombre no puede estar vacío.")
            continue

        if not nom.replace(" ", "").isalpha():
            print("Error: El nombre del candidato solo debe contener letras y espacios (sin números ni símbolos).")
            continue

        break  # Si el nombre es válido, salimos del bucle del nombre

    # 2. Bucle para validar obligatoriamente que el teléfono tenga exactamente 9 dígitos y solo números
    while True:
        telefon = input("Introduce el teléfono: ").strip()

        if not telefon.isdigit():
            print("Error: El teléfono no puede contener letras ni símbolos, solo números.")
            continue

        if len(telefon) != 9:
            print(f"Error: El teléfono debe tener exactamente 9 dígitos (has introducido {len(telefon)}).")
            continue

        break  # Si el teléfono es válido, salimos del bucle del teléfono

    # 3. Resto de inputs del formulario
    experiencia = input("Introduce los años de experiencia: ").strip()
    catalan = input("Introduce el nivel de catalán (0-3): ").strip()
    excel = input("Introduce el nivel de Excel (0-3): ").strip()
    titol = input("Introduce el nivel de titulación (0-3): ").strip()

    return nom, telefon, experiencia, catalan, excel, titol
```

```python
def validar_dades(nom, telefon, experiencia, catalan, excel, titol):
    # 1. Comprobar que el nombre y el teléfono no estén vacíos
    if not nom or not telefon:
        print("Error: El nombre y el teléfono no pueden estar vacíos.")
        return False

    # Comprobar que el nombre solo contenga letras y espacios (sin números)
    if not nom.replace(" ", "").isalpha():
        print("Error: El nombre del candidato solo debe contener letras y espacios (sin números ni símbolos).")
        return False

    # 2. Comprobar que sean dígitos válidos antes de convertir
    if not str(experiencia).replace('.', '', 1).isdigit() or not str(catalan).isdigit() or not str(excel).isdigit() or not str(titol).isdigit():
        print("Error: La experiencia, el nivel de catalán, Excel y titulación deben ser valores numéricos válidos.")
        return False

    # 3. Convertir a numérico para operar y validar rangos
    exp_num = float(experiencia)
    cat_num = int(catalan)
    exc_num = int(excel)
    tit_num = int(titol)

    if exp_num < 0:
        print("Error: Los años de experiencia no pueden ser negativos.")
        return False

    if not (0 <= cat_num <= 3 and 0 <= exc_num <= 3 and 0 <= tit_num <= 3):
        print("Error: Los niveles de catalán, Excel y titulación deben estar comprendidos entre 0 y 3.")
        return False

    return True
```

```python
def calcular_puntuacio(exp, cat, exc, tit):
    # Qué hace: Aplica una fórmula matemática a los datos ya validados.
    # Convertimos los textos a números para poder operar con ellos
    exp = float(exp)
    cat = int(cat)
    exc = int(exc)
    tit = int(tit)

    return (exp * 1.5) + cat + exc + tit
```

```python
def classificar_candidat(puntuacio):
    # Qué hace: Convierte un número en un concepto (una etiqueta).
    if puntuacio >= 12:
      return "Prioritat Alta"
    if puntuacio >= 8:
      return "Candidat adequat"
    else: return "No prioritari"
```

```python
def mostrar_individual(candidat):
    print(f"Candidat {candidat['nom']} registrat. Puntuació: {candidat['score']} ({candidat['prioritat']})")
```

```python
def mostrar_informe(candidats):
    print("\n--- INFORME FINAL DE SELECCIÓ ---")

    # Ordenar los candidatos: primero por la prioridad/score de forma descendente
    # Como 'score' ya refleja la puntuación, ordenamos directamente por score de mayor a menor (-c['score'])
    ordenats = sorted(candidats, key=lambda c: -c['score'])

    for c in ordenats:
        print(f"Prioritat: {c['classificacio']:<18} | Nom: {c['nom']:<15} | Tel: {c['telefon']:<10} | Punts: {c['score']}")
```

```python
def mostrar_informe(candidats):
    print("\n--- INFORME FINAL DE SELECCIÓ ---")

    # Mapeo para ordenar primero por prioridad y luego por puntuación descendente
    MAPA_PRIORITAT = {"Prioritat Alta": 0, "Candidat adequat": 1, "No prioritari": 2}

    ordenats = sorted(candidats, key=lambda x: (MAPA_PRIORITAT[x['prioritat']], -x['score']))

    for c in ordenats:
        print(f"{c['prioritat']:<18} | {c['nom']:<15} | Tel: {c['telefon']} | Punts: {c['score']}")
```

#Función principal

```python
def main_seleccio():
    candidats = []

    print("Iniciando proceso de selección. Máximo 10 candidatos válidos.")

    # El bucle continuará mientras tengamos menos de 10 candidatos válidos
    while len(candidats) < 10:
        datos = obtenir_dades()

        # Por si quieres conservar la opción manual de escribir 'salir'
        if datos[0].lower() == 'salir':
            break

        nom, telefon, experiencia, catalan, excel, titol = datos

        # Validar los datos introducidos
        if validar_dades(nom, telefon, experiencia, catalan, excel, titol):
            score = calcular_puntuacio(experiencia, catalan, excel, titol)
            classificacio = classificar_candidat(score)
            prioritat = classificacio

            candidat = {
                "nom": nom,
                "telefon": telefon,
                "score": score,
                "prioritat": prioritat,
                "classificacio": classificacio
            }

            candidats.append(candidat)
            mostrar_individual(candidat)

            print(f"--> Candidatos válidos inscritos: {len(candidats)}/10")
        else:
            print("Datos inválidos. Por favor, vuelve a introducir los datos correctamente.")

    # Al llegar a 10 candidatos, muestra el informe final automáticamente
    print("\n¡Se ha alcanzado el límite de 10 candidatos o se ha cerrado el proceso!")
    if candidats:
        mostrar_informe(candidats)
    else:
        print("No se han registrado candidatos válidos.")
```

#Funciones

```python
def obtenir_segment_i_factor(codigo_comportamiento):
    # Anem al diccionari de comportament i extraiem la informació
    dades = INFO_COMPORTAMENT[codigo_comportamiento]
    return dades["segment"], dades["factor"]
```

```python
def obtenir_factor_edat(grupo_edad):
    # Anem al diccionari d'edats i extraiem el factor
    return INFO_EDAT[grupo_edad]
```

```python
def calcular_risc(factor_codi, factor_edat):
    # Multipliquem els dos factors tal com diu la fórmula
    return round(factor_codi * factor_edat, 3)
```

```python
def processar_clients(llista_original):
    llista_final = []

    # Anem client per client
    for c in llista_original:
        # 1. Cridem les mini-funcions per obtenir les dades soltes
        nom_segment, f_codi = obtenir_segment_i_factor(c["codi_comportament"])
        f_edat = obtenir_factor_edat(c["grup_edat"])
        risc = calcular_risc(f_codi, f_edat)

        # 2. Obtenim el missatge directament del diccionari (SENSE CAP 'IF')
        txt_missatge = INFO_MISSATGES[c["codi_comportament"]]

        # 3. Creem la fitxa del client exactament amb els 10 camps demanats
        fitxa_client = {
            "id_cliente": c["id_client"],
            "nombre": c["nom"],
            "edad": c["edat"],
            "grupo_edad": c["grup_edat"],
            "codigo_comportamiento": c["codi_comportament"],
            "segment": nom_segment,
            "missatge": txt_missatge,
            "factor_codi": f_codi,
            "factor_edat": f_edat,
            "risc_final": risc
        }
        # Guardem la fitxa a la llista final
        llista_final.append(fitxa_client)

    return llista_final
```

```python
def mostrar_diccionari(clients_processats):
    # Ordenem els clients per risc de major a menor per veure'ls millor
    ordenats = sorted(clients_processats, key=lambda x: x["risc_final"], reverse=True)

    print("\n=== LLISTA DE CLIENTS PROCESSATS (ORDENATS PER RISC) ===")
    for c in ordenats:
        print(f"ID: {c['id_cliente']} | {c['nombre']} ({c['grupo_edad']}) -> Risc: {c['risc_final']} | Missatge: {c['missatge']}")
```

#Funcions

```python
def parsejar_pacient(registre_raw, diccionari_risc):
    """
    Transforma una línia de text "en brut" en un diccionari estructurat
    i calcula el seu risc total.
    """
    nom, edat, simptomes_raw = registre_raw.split("|")

    simptomes_nets = processar_simptomes(simptomes_raw)
    risc_total = calcular_risc_total(simptomes_nets, diccionari_risc)

    return {
        "nom": nom,
        "edat": int(edat),
        "simptomes": simptomes_nets,
        "risc_total": risc_total
    }
```

```python
def processar_simptomes(simptomes_string):
    """
    Agafa la cadena de símptomes separada per comes, la divideix i
    normalitza cada element utilitzant un bucle per netejar-los i afegir-los.
    """
    if not simptomes_string:
        return []

    # 1. Dividim correctament per comes
    llista_raw = simptomes_string.split(",")

    # 2. Netegem cada símptoma i el guardem a la nova llista
    llista_neta = []
    for s in llista_raw:
        s_neteixat = s.strip().lower().replace(" ", "_")
        llista_neta.append(s_neteixat)

    return llista_neta
```

```python
def calcular_risc_total(simptomes_normalitzats, diccionari_risc):
    """
    Calcula la suma dels valors de risc de cada símptoma.
    Si un símptoma no és al diccionari, compta com a 0.
    """
    total = 0
    for s in simptomes_normalitzats:
        total += diccionari_risc.get(s, 0)
    return total
```

#Orquestrador de las funciones

```python
def orquestrador_gestio_risc(llista_pacients, diccionari_risc):
    """
    Funció principal que coordina tot el procés: parseja,
    calcula el risc i ordena els pacients de major a menor prioritat.
    """
    # 1. Creem una llista buida per anar guardant els resultats
    pacients_estructurats = []

    # 2. Anem agafant cada text de pacient brut un per un
    for p in llista_pacients:
        # 3. El passem per la funció de parsejar
        pacient_parsejat = parsejar_pacient(p, diccionari_risc)
        # 4. El fiquem dins de la llista que havíem creat al principi
        pacients_estructurats.append(pacient_parsejat)

    # Ordenem de major a menor risc (descendent)
    pacients_ordenats = sorted(pacients_estructurats, key=lambda x: x["risc_total"], reverse=True)

    return pacients_ordenats
```
