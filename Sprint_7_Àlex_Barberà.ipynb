```python
# Mapeo de prioridades para ordenación (Prioridad Alta = 0, Adecuado = 1, No prioritario = 2)
MAPA_PRIORITAT = {"Prioritat Alta": 0, "Candidat adequat": 1, "No prioritari": 2}
```

#Funciones

Paso 1: Entrada y extracción: obtenir_dades (prepara el terreno).

Paso 2: Control de calidad: validar_dades (filtro de seguridad).

Paso 3: Lógica de negocio: calcular_puntuacio (aplica las reglas de la empresa).

Paso 4: Toma de decisiones: classificar_candidat (traduce resultados a conclusiones).


```python
def obtenir_dades(test_data, index):
    # El * 6 está ahí porque cada candidato tiene exactamente 6 datos en esa lista.
    # Se hace + 6 porque, si cada candidato ocupa 6 posiciones, necesitamos contar 6 lugares desde donde empezó para saber dónde termina.
    inicio = index * 6
    final = inicio + 6

    # Creamos una lista vacía
    candidato = []

    # Vamos añadiendo elemento a elemento
    for i in range(inicio, final):
        candidato.append(test_data[i])

    return candidato
    # Esta funcion aisla los candidatos.
    # Por ejemplo: ["Clara Gómez", "655221199", 3, 3, 3, 2] en vez de tener un listado enorme con toooodos los candidatos.
```

```python
def validar_dades(experiencia, catalan, excel, titol):
    # Creamos una lista con los valores a comprobar
    # Qué hace: Actúa como un portero. Antes de hacer cálculos, se asegura de que los datos recibidos tengan sentido.
    mis_valores = [catalan, excel, titol]

    # Recorremos cada valor uno a uno
    for v in mis_valores:
        # Si encontramos UNO solo que NO cumpla la condición,
        # ya sabemos que todo el conjunto es inválido.
        if v < 0 or v > 3:
            return False

    # Si el bucle termina sin haber encontrado errores, significa que todos son correctos
    return True
```

```python
def calcular_puntuacio(exp, cat, exc, tit):
    # Qué hace: Aplica una fórmula matemática a los datos ya validados.
    # Lógica: Aquí es donde entra la "ponderación". Observa que exp (experiencia) se multiplica por 1.5, dándole más peso en la nota final que el resto de habilidades.
    return (exp * 1.5) + cat + exc + tit
```

```python


def classificar_candidat(puntuacio):
    # Qué hace: Convierte un número en un concepto (una etiqueta).
    if puntuacio >= 12: return "Prioritat Alta"
    if puntuacio >= 8: return "Candidat adequat"
    return "No prioritari"
```

#Función principal

```python
def main_seleccio(inputs_test):
    candidats = []
    i = 0

    # Procesar máximo 10 candidatos
    while len(candidats) < 10 and (i * 6) < len(inputs_test):
        nom, tel, exp, cat, exc, tit = obtenir_dades(inputs_test, i)

        if validar_dades(exp, cat, exc, tit):
            score = calcular_puntuacio(exp, cat, exc, tit)
            clas = classificar_candidat(score)

            candidat = {"nom": nom, "tel": tel, "score": score, "prioritat": clas}
            candidats.append(candidat)
            print(f"Candidat {nom} registrat. Puntuació: {score} ({clas})")
        else:
            print(f"Error en dades del candidat {nom}: Dades no vàlides.")
        i += 1

    # Informe final
    print("\n--- INFORME FINAL DE SELECCIÓ ---")
    # Ordenar por el valor numérico de la prioridad y luego por score descendente
    ordenats = sorted(candidats, key=lambda x: (MAPA_PRIORITAT[x['prioritat']], -x['score']))

    for c in ordenats:
        print(f"{c['prioritat']:<18} | {c['nom']:<15} | Tel: {c['tel']} | Punts: {c['score']}")
```

```python
# Un diccionari de diccionaris per saber el text del segment i el factor de comportament
INFO_COMPORTAMENT = {
    0: {"segment": "Navega pero no añade productos", "factor": 0.10},
    1: {"segment": "Añade productos pero no llega al checkout", "factor": 0.25},
    2: {"segment": "Llega al checkout pero no paga", "factor": 0.50},
    3: {"segment": "Carrito de compra abandonado hace < 24 h", "factor": 0.75},
    4: {"segment": "Carrito de compra abandonado hace > 24 h", "factor": 0.90}
}

# Un diccionari per saber el factor de l'edat
INFO_EDAT = {
    "18-25": 1.20,
    "26-40": 1.00,
    "41-60": 0.85,
    "60+": 0.70
}
```

#Funciones

```python
def obtenir_segment_i_factor(codigo_comportamiento):
    # Anem al diccionari de comportament i extraiem la informació
    dades = INFO_COMPORTAMENT[codigo_comportamiento]
    return dades["segment"], dades["factor"]
```

```python
def obtenir_factor_edat(grupo_edad):
    # Anem al diccionari d'edats i extraiem el factor
    return INFO_EDAT[grupo_edad]
```

```python
def calcular_risc(factor_codi, factor_edat):
    # Multipliquem els dos factors tal com diu la fórmula
    return round(factor_codi * factor_edat, 3)
```

```python
def processar_clients(llista_original):
    llista_final = []

    # Anem client per client
    for c in llista_original:
        # 1. Cridem les mini-funcions per obtenir les dades soltes
        nom_segment, f_codi = obtenir_segment_i_factor(c["codi_comportament"])
        f_edat = obtenir_factor_edat(c["grup_edat"])
        risc = calcular_risc(f_codi, f_edat)

        # 2. Obtenim el missatge directament del diccionari (SENSE CAP 'IF')
        txt_missatge = INFO_MISSATGES[c["codi_comportament"]]

        # 3. Creem la fitxa del client exactament amb els 10 camps demanats
        fitxa_client = {
            "id_cliente": c["id_client"],
            "nombre": c["nom"],
            "edad": c["edat"],
            "grupo_edad": c["grup_edat"],
            "codigo_comportamiento": c["codi_comportament"],
            "segment": nom_segment,
            "missatge": txt_missatge,
            "factor_codi": f_codi,
            "factor_edat": f_edat,
            "risc_final": risc
        }
        # Guardem la fitxa a la llista final
        llista_final.append(fitxa_client)

    return llista_final
```

```python
def mostrar_diccionari(clients_processats):
    # Ordenem els clients per risc de major a menor per veure'ls millor
    ordenats = sorted(clients_processats, key=lambda x: x["risc_final"], reverse=True)

    print("\n=== LLISTA DE CLIENTS PROCESSATS (ORDENATS PER RISC) ===")
    for c in ordenats:
        print(f"ID: {c['id_cliente']} | {c['nombre']} ({c['grupo_edad']}) -> Risc: {c['risc_final']} | Missatge: {c['missatge']}")
```

        // 2. Normalizar síntomas

        // 4. Guardar objeto estructurado

    RETORNAR lista_pacientes_procesados

// PASO 3: EJECUCIÓN Y ORDENACIÓN

# Diccionaris inicials de l'exercici


```python
pacients = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]

nivells_risc = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}
```

#Funcions

```python
def parsejar_pacient(registre_raw, diccionari_risc):
    """
    Transforma una línia de text "en brut" en un diccionari estructurat
    i calcula el seu risc total.
    """
    nom, edat, simptomes_raw = registre_raw.split("|")

    simptomes_nets = processar_simptomes(simptomes_raw)
    risc_total = calcular_risc_total(simptomes_nets, diccionari_risc)

    return {
        "nom": nom,
        "edat": int(edat),
        "simptomes": simptomes_nets,
        "risc_total": risc_total
    }
```

```python
def processar_simptomes(simptomes_string):
    """
    Agafa la cadena de símptomes separada per comes, la divideix i
    normalitza cada element utilitzant un bucle per netejar-los i afegir-los.
    """
    if not simptomes_string:
        return []

    # 1. Dividim correctament per comes
    llista_raw = simptomes_string.split(",")

    # 2. Netegem cada símptoma i el guardem a la nova llista
    llista_neta = []
    for s in llista_raw:
        s_neteixat = s.strip().lower().replace(" ", "_")
        llista_neta.append(s_neteixat)

    return llista_neta
```

```python
def calcular_risc_total(simptomes_normalitzats, diccionari_risc):
    """
    Calcula la suma dels valors de risc de cada símptoma.
    Si un símptoma no és al diccionari, compta com a 0.
    """
    total = 0
    for s in simptomes_normalitzats:
        total += diccionari_risc.get(s, 0)
    return total
```

#Orquestrador de las funciones

```python
def orquestrador_gestio_risc(llista_pacients, diccionari_risc):
    """
    Funció principal que coordina tot el procés: parseja,
    calcula el risc i ordena els pacients de major a menor prioritat.
    """
    # 1. Creem una llista buida per anar guardant els resultats
    pacients_estructurats = []

    # 2. Anem agafant cada text de pacient brut un per un
    for p in llista_pacients:
        # 3. El passem per la funció de parsejar
        pacient_parsejat = parsejar_pacient(p, diccionari_risc)
        # 4. El fiquem dins de la llista que havíem creat al principi
        pacients_estructurats.append(pacient_parsejat)

    # Ordenem de major a menor risc (descendent)
    pacients_ordenats = sorted(pacients_estructurats, key=lambda x: x["risc_total"], reverse=True)

    return pacients_ordenats
```

# --- Ejecución del orquestrador ---

```python
resultat_final = orquestrador_gestio_risc(pacients, nivells_risc)
```
