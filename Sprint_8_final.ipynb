**Nota: Primero de todo, cargamos el Dataset y corregimos el problema de los encabezados, para poder filtrar a posteriori por columna cuántos DNI hay repetidos y demás.**

```python
import pandas as pd

# Cargamos el Excel localmente con skiprows=3 y el motor openpyxl
df = pd.read_excel(
    "sprint8_complex.xlsx", skiprows=3, engine="openpyxl"
)
df = df.drop(columns=["Unnamed: 0"])

# Convertir de manera masiva cualquier texto mal codificado
for col in df.select_dtypes(include=["object"]).columns:
  df[col] = (
      df[col]
      .astype(str)
      .apply(
          lambda x: (
              x.encode("latin1", errors="ignore")
              .decode("utf-8", errors="ignore")
              if "Ã" in x or "Ô" in x
              else x
          )
      )
  )

display(df.head(10))
```

**Ahora sí, comprobamos cuantos DNI hay repetidos.**

```python
# 1. Contamos cuantas veces sale cada DNI
conteo = df['DNI'].value_counts()

# 2. Filtramos solo los que se repiten (> 1)
duplicats = conteo[conteo > 1]

# 3. Imprime el total y cuales son
print(f"Nombre de DNI repetits: {len(duplicats)}")
print("Els DNI que es repeteixen són:")
print(duplicats)

print("-" * 50) # Una linea separadora visual

# 4. Extraer e imprimir
dni_repetits = duplicats.index
files_duplicades = df[df['DNI'].isin(dni_repetits)].sort_values(by='DNI')

print("-------- Listado de DNIs duplicados --------")
display(files_duplicades)
```

**» Calcular un increment salarial total combinant:**

Grup professional.

Situació familiar.

Cost de vida de la ciutat.


```python
import pandas as pd
import json

# 1. Cargar la política desde el archivo JSON
with open('/content/drive/MyDrive/Pandas_Sprint_8/politica_salarial.json', 'r', encoding='utf-8') as f:
    politica = json.load(f)

# 2. Preparar los datos necesarios para el cálculo
edad = 2026 - pd.to_numeric(df_enriquit['Any_de_Naixement'], errors='coerce')
franja = pd.cut(edad, bins=[0, 29, 44, 59, 120], labels=['18-29', '30-44', '45-59', '60+'])
coste = pd.cut(df_enriquit['Cost_vida_index'].fillna(df_enriquit['Cost_vida_index'].mean()), bins=[0, 65, 75, 100], labels=['low', 'medium', 'high'])
fills_num = pd.to_numeric(df_enriquit['Nombre_fills'], errors='coerce').fillna(0)

# 3. Calcular el incremento total sumando cada factor convertido explícitamente a float
df_enriquit['Increment_Salari'] = (
    df_enriquit['Grup_Professional'].map(politica['base_increment_by_group']).fillna(0).astype(float) +
    (fills_num > 0).map({True: politica['children_adjustment']['has_children'], False: politica['children_adjustment']['no_children']}).astype(float) +
    franja.map(politica['age_adjustment']).fillna(0).astype(float) +
    coste.map(politica['cost_of_living_adjustment']).fillna(0).astype(float)
)

# 4. Aplicar el tope máximo (cap) definido en las reglas del JSON
max_tope = politica['rules']['max_total_increment_percent']
df_enriquit['Increment_Salari'] = df_enriquit['Increment_Salari'].clip(upper=max_tope)

# 5. Formatear la columna para que muestre el símbolo '%' al lado
df_enriquit['Increment_Salari'] = df_enriquit['Increment_Salari'].apply(lambda x: f"{x}%")

# 6. Visualizar las primeras 10 filas de todo el DataFrame
df_enriquit.head(40)
```

```python
# Aquí limpiamos el símbolo de euro, eliminamos los puntos de miles y convertimos a entero
df_enriquit['Salari_mensual'] = (
    df_enriquit['Salari_mensual']
    .astype(str)
    .str.replace('€', '', regex=False)
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
)
df_enriquit['Salari_mensual'] = pd.to_numeric(df_enriquit['Salari_mensual'], errors='coerce').astype('Int64')

# Calcular la ratio (redondeada a 2 decimales)
df_enriquit['Ratio_Salari_CostVida'] = (df_enriquit['Salari_mensual'] / df_enriquit['Cost_vida_index']).round(2)

# Filtrar para eliminar los registros incompletos (donde aparezcan valores NA)
df_filtrat_ratio = df_enriquit.dropna(subset=['Ratio_Salari_CostVida'])

# Muestra
display(df_filtrat_ratio[['Nom', 'Cognoms', 'Salari_mensual', 'Cost_vida_index', 'Ratio_Salari_CostVida']].head(20))
```

**La seva utilitat analítica.**

Con el ratio de Salario / Coste de vida obtenemos el índex relativo de poder adquisitivo. Por lo que, podemos saber el nivel de vida de un individuo en base a su remuneración y su vivienda actual (coste de vida).



**Les decisions preses.**

**Limpieza y estandarización de cadenas monetarias:** Se ha transformado la columna Salari_mensual a texto para eliminar de forma segura el símbolo de euro (€), los puntos de miles y comas.

**Filtrado de registros incompletos (dropna):** Se ha aplicado un filtro de eliminación de filas basado en la columna de la ratio para descartar automáticamente aquellos registros donde faltaban datos de origen (NA), evitando la propagación de errores en la tabla final.

**Visualización:** Se ha utilizado la función display() mostrando únicamente las primeras 20 filas de las columnas clave (Nom, Cognoms, Salari_mensual, Cost_vida_index y Ratio_Salari_CostVida) para comprobar visualmente con la máxima claridad los datos obtenidos (sin un exceso de datos).


**2. Indicador de dependència familiar (Salari / Nombre de fills)**

**Fórmula principal:** Indicador Dependencia Familiar = Salari Mensual / Nombre Fills

```python
# Convertir las columnas a tipo numérico por si contienen valores de texto
df_enriquit['Nombre_fills'] = pd.to_numeric(
    df_enriquit['Nombre_fills'], errors='coerce'
)
df_enriquit['Salari_mensual'] = pd.to_numeric(
    df_enriquit['Salari_mensual'], errors='coerce'
)

# Creación de la variable derivada sumando 1 al denominador y redondeando a 2 decimales
df_enriquit['Indicador_Dependencia_Familiar'] = (
    df_enriquit['Salari_mensual'] / (df_enriquit['Nombre_fills'] + 1)
).round(2)

# Mostrar una pequeña muestra para comprobar el resultado
display(
    df_enriquit[
        [
            'Nom',
            'Cognoms',
            'Salari_mensual',
            'Nombre_fills',
            'Indicador_Dependencia_Familiar',
        ]
    ].head(20)
)
```

**La seva utilitat analítica.**

El indicador de dependencia familiar permite evaluar la capacidad adquisitiva real y la presión económica individualizada dentro del hogar, poniendo en relación el salario mensual con el número de cargas familiares reales.

Es decir, facilita la detección de perfiles que, a pesar de recibir un salario moderado, sufren un riesgo elevado de estrés económico debido a un volumen elevado de descendencia

**Les decisions preses.**

**Inclusión de la unidad en el denominador (+ 1):** Se ha sumado una unidad al número de hijos (Nombre_fills + 1) como parte de la fórmula de reparto. Esta decisión responde a una doble finalidad.

- Evitar errores técnicos: Previene de forma nativa cualquier posible división por cero en aquellos individuos o perfiles que no cuentan con descendencia.

- Reflejar la unidad real: Contempla a la propia persona trabajadora dentro de la estructura familiar. De este modo, si un empleado tiene un único hijo, el salario se divide entre dos, reflejando que los recursos económicos deben cubrir tanto al proveedor familiar asó como a la persona dependiente.



**3. Índex de mobilitat: Km anuals / Temperatura mitjana**

**Fórmula:** Index Mobilidad = Km Anuales / temperatura Mediana Ciudad

```python
# 1. Forzamos la conversión a numérico de ambas columnas por si son de tipo string
df_enriquit['Km_anuals'] = pd.to_numeric(
    df_enriquit['Km_anuals'], errors='coerce'
)
df_enriquit['Temperatura_mitjana_ciutat'] = pd.to_numeric(
    df_enriquit['Temperatura_mitjana_ciutat'], errors='coerce'
)

# Eliminar filas duplicadas basandome en el DNI
df_enriquit = df_enriquit.drop_duplicates(subset=['DNI'], keep='first')

# 2. Creación de la operación aritmética
df_enriquit['Index_Mobilitat'] = (
    df_enriquit['Km_anuals'] / df_enriquit['Temperatura_mitjana_ciutat']
).round(2)

# 3. Filtrar para eliminar los registros que se hayan quedado sin valor
df_filtrat_mob = df_enriquit.dropna(subset=['Index_Mobilitat'])

# 4. Resultado
display(
    df_filtrat_mob[
        [
            'Nom',
            'Cognoms',
            'Km_anuals',
            'Temperatura_mitjana_ciutat',
            'Index_Mobilitat',
        ]
    ].head(20)
)
```

**La seva utilitat analítica.**

Este valor permite normalizar el esfuerzo de desplazamiento teniendo en cuenta el factor climático, facilitando la comparación de la movilidad entre trabajadores independientemente de la ciudad en la que residan.

**Les decisions preses.**

**Operación directa y limpieza de valores nulos (dropna):** Dado que no se parte de temperaturas medias iguales a cero, se ha optado por una división directa y se ha aplicado un método .dropna(subset=['Index_Mobilitat']). Esta decisión asegura que el dataset contenga registros completos, eliminando las filas incompletas para así disponer de una muestra fiable para su interpretación.

**Redondeo:** Se ha limitado el resultado final a dos decimales (.round(2)) para facilitar la lectura y comparativa entre los registros.
